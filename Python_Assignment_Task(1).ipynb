## Task 1: Mobile Shop Sales & Tax Calculator

**Problem Statement:**

Suppose a mobile phone shop has 5 sales employees. Write a Python program that uses a `for` loop to enter the daily sales amount (in Rs.) generated by each employee. Calculate the tax using the following rules:
- Sales < Rs. 50,000  →  **5% Tax**
- Rs. 50,000 ≤ Sales ≤ Rs. 100,000  →  **10% Tax**
- Sales > Rs. 100,000  →  **15% Tax**

**Required Output:**
1. Sales Amount and Tax Amount for each employee.
2. Total Sales and Average Sales for all 5 employees.

```python
# ==============================================================================
# Task 1: Mobile Shop Sales & Tax Calculator
# Description: Calculates individual tax amounts, total sales, and average sales 
#              for 5 employees based on specified tax slabs.
# ==============================================================================

# Variables initialization to store aggregated results
total_sales = 0
num_employees = 5

print("==================================================")
print("      MOBILE SHOP DAILY SALES & TAX REPORT        ")
print("==================================================\n")

# Loop to collect data and perform calculations for each employee
for i in range(1, num_employees + 1):
    sales = float(input(f"Enter daily sales amount for Employee {i} (in Rs.): "))
    
    # Tax Calculation Logic
    if sales < 50000:
        tax_rate = 0.05
    elif 50000 <= sales <= 100000:
        tax_rate = 0.10
    else:  # sales > 100000
        tax_rate = 0.15
        
    tax_amount = sales * tax_rate
    total_sales += sales
    
    # Display individual report
    print(f"--- Employee {i} Summary ---")
    print(f"Sales Amount : Rs. {sales:,.2f}")
    print(f"Tax Amount   : Rs. {tax_amount:,.2f} ({int(tax_rate * 100)}% tax rate)")
    print("-" * 30 + "\n")

# Calculate overall statistics
average_sales = total_sales / num_employees

# Display overall shop summary
print("==================================================")
print("              OVERALL SHOP SUMMARY                ")
print("==================================================")
print(f"Total Sales   : Rs. {total_sales:,.2f}")
print(f"Average Sales : Rs. {average_sales:,.2f}")
print("==================================================")
```

## Output Explanation:
The program successfully takes input for 5 employees, calculates tax based on specified slabs (5%, 10%, 15%), and prints individual summaries along with the overall Total Sales and Average Sales.

## Task 2: NumPy Array Manipulation

**Problem Statement:**

Write a Python program using NumPy to:
1. Create a NumPy array containing the marks `[78, 85, 92, 67, 88]`.
2. Display the original array.
3. Convert the array into float data type.
4. Display the converted array.

```python
# ==============================================================================
# Task 2: NumPy Array Creation & Data Type Conversion
# Description: Demonstrates creating a NumPy array, inspecting it, converting 
#              its data type to float, and displaying the transformed array.
# ==============================================================================

import numpy as np

print("==================================================")
print("          NUMPY ARRAY MANIPULATION                ")
print("==================================================\n")

# 1. Create a NumPy array containing the given marks
marks_list = [78, 85, 92, 67, 88]
marks_array = np.array(marks_list)

# 2. Display the original array along with its data type
print("1. Original NumPy Array:")
print(f"   Array     : {marks_array}")
print(f"   Data Type : {marks_array.dtype}\n")

# 3. Convert the array into float data type
float_marks_array = marks_array.astype(float)

# 4. Display the converted array along with its new data type
print("2. Converted Array (Float Data Type):")
print(f"   Array     : {float_marks_array}")
print(f"   Data Type : {float_marks_array.dtype}\n")

print("==================================================")
```

## Output Explanation:
The NumPy array is successfully created with default integer data type int64 (or int32), and then converted into floating-point numbers using .astype(float) where values now include decimal points.

## Task 3: Bank Account System using Object-Oriented Programming (OOP)

**Problem Statement:**
Design a simple Bank Account System using Object-Oriented Programming (OOP) in Python using `Customer` and `Account` classes.

**Implemented Functionalities:**
1. Create a customer account.
2. Deposit money into the account.
3. Withdraw money with insufficient balance checking.
4. Display the current balance and account details.


```python
class Customer:
    def __init__(self, name):
        self.name = name

class Account:
    def __init__(self, acc_no, customer, balance=0):
        self.acc_no = acc_no
        self.customer = customer
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited: Rs. {amount}")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Withdrawn: Rs. {amount}")
        else:
            print("Insufficient balance!")

    def display_balance(self):
        print(f"Current Balance: Rs. {self.balance}")


# Testing the System
cust = Customer("Ayesha")
acc = Account("ACC123", cust, 25000)

acc.deposit(15000)
acc.withdraw(8000)
acc.display_balance()
```

## Output Explanation:
The OOP system creates a Customer instance linked to an Account object. It successfully demonstrates depositing Rs. 15,000, withdrawing Rs. 8,000, handling overdraft error checking for excessive withdrawal, and displaying the final balance of Rs. 32,000.
